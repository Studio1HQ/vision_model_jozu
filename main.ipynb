### Demo: Visual question answering on a sample image

```python
import base64
from openai import OpenAI

# ── 1. Connect to your local llama-server ──────────────────────────────────
client = OpenAI(
    base_url="http://localhost:8000/v1",
    api_key="sk-not-required"
)
```

```python
# ── 2. Helper: encode image to base64 ─────────────────────────────────────
def encode_image(image_path: str) -> str:
    with open(image_path, "rb") as f:
        return base64.b64encode(f.read()).decode("utf-8")
```

```python
# ── 3. encode sample  ─────────────────────────────────────────────────
image_path = "receipt.jpg"
b64_encoded_image = encode_image(image_path)
```

```python
# ── 4. query the VLM with the image  ─────────────────────────────────────────────────
response = client.chat.completions.create(
    model="qwen3.6-35b-a3b-gguf",
    messages=[
        {
            "role": "user",
            "content": [
                {
                    "type": "image_url",
                    "image_url": {
                        "url": f"data:image/jpeg;base64,{b64_encoded_image}"
                    }
                },
                {
                    "type": "text",
                    "text": "Describe the image in detail."
                }
            ]
        }
    ],
    max_tokens=8192
)

model_output = response.choices[0].message.content
print("Model output:", model_output)
```
